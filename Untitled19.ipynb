```python
df = df.dropna(subset=['Phone'])
print(f"Rows after dropping missing phones: {df.shape[0]}")
```

```python
df['Source'] = df['Source'].fillna('Unknown')
print(df['Source'].value_counts())
```

```python
df = df.drop(columns=['Email', 'Address'])
print(f"Remaining columns: {list(df.columns)}")
```

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Duplicate phones: {df['Phone'].duplicated().sum()}")
```

```python
dupes = df[df['Phone'].duplicated(keep=False)].sort_values('Phone')
print(f"Total rows involved: {len(dupes)}")
print(dupes[['Phone', 'Postcode', 'Total Orders', 'Last Order Date', 'Avg Order Value (£)', 'Order Type']].head(20))
```

```python
df['Phone'] = df['Phone'].astype(str).str.replace('.0', '', regex=False)
short_phones = df[df['Phone'].str.len() < 10]
print(f"Rows with less than 10 digit phone: {len(short_phones)}")
print(f"Out of total: {len(df)}")
print(f"Percentage: {round(len(short_phones)/len(df)*100, 1)}%")
```

```python
print("Phone number length distribution:")
print(df['Phone'].str.len().value_counts().sort_index())
```

```python
# Fix 9-digit phones by adding leading 0
df.loc[df['Phone'].str.len() == 9, 'Phone'] = '0' + df.loc[df['Phone'].str.len() == 9, 'Phone']

# Drop rows with phone less than 10 digits (1-3 and 7-8 digit ones)
df = df[df['Phone'].str.len() >= 10]

print(f"Rows remaining: {len(df)}")
print(f"Duplicate phones now: {df['Phone'].duplicated().sum()}")
print(f"\nPhone length distribution:")
print(df['Phone'].str.len().value_counts().sort_index())
```

```python
dupes = df[df['Phone'].duplicated(keep=False)].sort_values('Phone')
print(f"Total rows involved: {len(dupes)}")
print(dupes[['Phone', 'Postcode', 'Total Orders', 'Last Order Date', 'Avg Order Value (£)']].head(20))
```

```python
df['Last Order Date'] = pd.to_datetime(df['Last Order Date'])
df = df.sort_values('Last Order Date', ascending=False).drop_duplicates(subset='Phone', keep='first')
print(f"Rows remaining: {len(df)}")
print(f"Duplicate phones: {df['Phone'].duplicated().sum()}")
```

```python
print(df['Total Orders'].describe())
print(df['Avg Order Value (£)'].describe())
print(f"Customers with 0 orders: {(df['Total Orders'] == 0).sum()}")
print(f"Customers with £0 value: {(df['Avg Order Value (£)'] == 0).sum()}")
```

```python
df = df[(df['Total Orders'] > 0) & (df['Avg Order Value (£)'] > 0)]
print(f"Rows remaining: {len(df)}")
```

```python
df['Customer_ID'] = ['CUST_' + str(i).zfill(4) for i in range(1, len(df) + 1)]
df = df.drop(columns=['Phone', 'ID'])
print(df.head())
print(f"\nFinal dataset: {len(df)} customers, {len(df.columns)} columns")
print(f"Columns: {list(df.columns)}")
```
